# Stage 1: Data Cleaning & Preprocessing

Replication study based on Ahmad et al. (2020) - *Fake News Detection Using Machine Learning Ensemble Methods*.

## Objectives
1. **DS1 (ISOT)**: Load `Fake.csv` (label 0) and `True.csv` (label 1). Merge headline (`title`) and body (`text`).
2. **DS2 (Kaggle)**: Load `train.csv`. Standardize labels to `0` for Fake, `1` for Truthful. Merge headline (`title`) and body (`text`). Ignore unlabelled `test.csv`.
3. **DS3 (Kaggle Detection)**: Load `data.csv`. Standardize labels (`FAKE` -> 0, `REAL` -> 1). Merge headline (`Headline`) and body (`Body`).
4. **Text Cleaning**: Strip HTML tags, remove non-content metadata columns (`author`, `date`, `URLs`, `subject`), and filter articles where combined body length < 20 words.
5. **Exporting**: Save `DS1_cleaned.csv`, `DS2_cleaned.csv`, and `DS3_cleaned.csv` to `data/processed/`.
6. **Master DS4**: Concatenate DS1, DS2, DS3, shuffle with `random_state=42`, and save `DS4_combined.csv`.
